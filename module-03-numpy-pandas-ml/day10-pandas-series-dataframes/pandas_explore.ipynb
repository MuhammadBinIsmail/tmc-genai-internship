# Day 10 — Pandas I: Series & DataFrames
Loading and exploring tabular data with Pandas.
Dataset: Titanic passenger data

## 1. Loading Data & Series vs DataFrame

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# A Series is a single column; a DataFrame is the full table
fare_series = df["Fare"]
print(type(fare_series))
print(type(df))
df.head()
```

## 2. Inspecting the Data

## 3. Selection with loc and iloc

```python
# loc: label-based selection
print(df.loc[0:4, ["Name", "Age", "Survived"]])

# iloc: position-based selection
print(df.iloc[0:5, 0:4])
```

## 4. Filtering Rows

```python
survivors = df[df["Survived"] == 1]
first_class = df[df["Pclass"] == 1]
young_survivors = df[(df["Survived"] == 1) & (df["Age"] < 18)]

print("Survivors:", len(survivors))
print("First class passengers:", len(first_class))
print("Young survivors (under 18):", len(young_survivors))
```

## 5. Adding Columns

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = df["FamilySize"] == 1

df[["Name", "FamilySize", "IsAlone"]].head()
```

## 6. Sorting

```python
oldest_passengers = df.sort_values("Age", ascending=False).head(5)
oldest_passengers[["Name", "Age", "Survived"]]
```

## 7. Ten Insights from the Data

```python
insight_1 = df["Survived"].mean()
print(f"1. Overall survival rate: {insight_1:.1%}")

insight_2 = df.groupby("Sex")["Survived"].mean()
print(f"\n2. Survival rate by sex:\n{insight_2}")

insight_3 = df.groupby("Pclass")["Survived"].mean()
print(f"\n3. Survival rate by class:\n{insight_3}")

insight_4 = df["Age"].mean()
print(f"\n4. Average passenger age: {insight_4:.1f} years")

insight_5 = df["Age"].isna().sum()
print(f"\n5. Missing age values: {insight_5} out of {len(df)}")

insight_6 = df["Fare"].max()
top_fare_passenger = df.loc[df["Fare"].idxmax(), "Name"]
print(f"\n6. Highest fare paid: ${insight_6:.2f} by {top_fare_passenger}")

insight_7 = df["Embarked"].value_counts()
print(f"\n7. Passengers by embarkation port:\n{insight_7}")

insight_8 = df[df["IsAlone"]]["Survived"].mean()
insight_8b = df[~df["IsAlone"]]["Survived"].mean()
print(f"\n8. Survival rate - alone: {insight_8:.1%} vs with family: {insight_8b:.1%}")

insight_9 = df["FamilySize"].max()
largest_family_name = df.loc[df["FamilySize"].idxmax(), "Name"]
print(f"\n9. Largest family size aboard: {insight_9} ({largest_family_name})")

insight_10 = df.groupby("Pclass")["Fare"].mean()
print(f"\n10. Average fare by class:\n{insight_10}")
```
