# Day 9 — NumPy Fundamentals
Use NumPy for fast numerical computing.

## 1. ndarray vs Python Lists

```python
import numpy as np

python_list = [1, 2, 3, 4, 5]
numpy_array = np.array(python_list)

print("List:", python_list, type(python_list))
print("Array:", numpy_array, type(numpy_array))
print("dtype:", numpy_array.dtype)
```

## 2. dtypes & Shapes

```python
int_array = np.array([1, 2, 3], dtype=np.int32)
float_array = np.array([1.5, 2.5, 3.5], dtype=np.float64)

matrix = np.array([[1, 2, 3], [4, 5, 6]])

print("int_array dtype:", int_array.dtype)
print("float_array dtype:", float_array.dtype)
print("matrix shape:", matrix.shape)
print("matrix ndim:", matrix.ndim)
print("matrix size:", matrix.size)
```

## 3. Creating Arrays

```python
zeros = np.zeros((2, 3))
ones = np.ones((2, 3))
range_array = np.arange(0, 10, 2)
linspace_array = np.linspace(0, 1, 5)
random_array = np.random.rand(3, 3)

print("zeros:\n", zeros)
print("ones:\n", ones)
print("arange:", range_array)
print("linspace:", linspace_array)
print("random:\n", random_array)
```

## 4. Indexing, Slicing & Boolean Masks

```python
grid = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

print("Element at row 1, col 2:", grid[1, 2])
print("First two rows:\n", grid[:2])
print("Second column:", grid[:, 1])

mask = grid > 50
print("Boolean mask:\n", mask)
print("Values greater than 50:", grid[mask])
```

## 5. Vectorized Operations & Broadcasting

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print("Element-wise sum:", a + b)
print("Element-wise product:", a * b)

# Broadcasting: adding a scalar to every element without a loop
print("Broadcast add 5:", a + 5)

matrix = np.array([[1, 2, 3], [4, 5, 6]])
row_vector = np.array([10, 20, 30])
print("Broadcast matrix + row vector:\n", matrix + row_vector)
```

## 6. Aggregations

```python
data = np.array([[1, 2, 3], [4, 5, 6]])

print("Sum (all):", data.sum())
print("Sum (axis=0, per column):", data.sum(axis=0))
print("Sum (axis=1, per row):", data.sum(axis=1))
print("Mean:", data.mean())
print("Std dev:", data.std())
```

## 7. Reshaping & Stacking

```python
flat = np.arange(1, 7)
reshaped = flat.reshape(2, 3)

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print("Reshaped:\n", reshaped)
print("Vertical stack:\n", np.vstack([x, y]))
print("Horizontal stack:", np.hstack([x, y]))
```

## 8. Basic Linear Algebra

```python
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

matrix_a = np.array([[1, 2], [3, 4]])
matrix_b = np.array([[5, 6], [7, 8]])

print("Dot product:", np.dot(v1, v2))
print("Matrix multiplication:\n", matrix_a @ matrix_b)
print("Transpose:\n", matrix_a.T)
```

## 9. Hands-on: Vectorization Speed Comparison
Comparing a pure-Python loop against NumPy's vectorized operations on the same computation.

```python
import time

size = 1_000_000
data = list(range(size))
numpy_data = np.arange(size)

# Pure Python loop
start = time.time()
squared_python = [x ** 2 for x in data]
python_time = time.time() - start

# NumPy vectorized
start = time.time()
squared_numpy = numpy_data ** 2
numpy_time = time.time() - start

print(f"Pure Python time: {python_time:.4f} seconds")
print(f"NumPy vectorized time: {numpy_time:.4f} seconds")
print(f"NumPy is {python_time / numpy_time:.1f}x faster")
```
